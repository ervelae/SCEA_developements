```python
# Define ellipse function with rotation
def ellipse(x, y, a=2, b=1, angle=0):
    cos_angle = np.cos(angle)
    sin_angle = np.sin(angle)
    x_rot = cos_angle * x + sin_angle * y
    y_rot = -sin_angle * x + cos_angle * y
    return (x_rot**2) / (a**2) + (y_rot**2) / (b**2) - 1  # Inside ellipse if <= 0

a = 2
b = 1
angle = np.pi / 6  # 30 degrees

point1 = (1.0, 0.0)
point2 = (2.0, 1.5)
point3 = (0.0, 0.0)

print(ellipse(point1[0], point1[1], a, b, angle))  # True
print(ellipse(point2[0], point2[1], a, b, angle))  # False
print(ellipse(point3[0], point3[1], a, b, angle))  # True
```

# Half-ellipse

```python
def half_ellipse(X,Y, a=2, b=1, rotation=0, rotation_unit='radians'):
    """
    Returns the value of the half-ellipse function at points (X, Y).
    Inside half-ellipse if value <= 1.
    Parameters:
    X : np.ndarray
        X-coordinates of the points.
    Y : np.ndarray
        Y-coordinates of the points.
    a : float
        Axis length towards the rotation direction.
    b : float
        Axis length perpendicular to the rotation direction.
    rotation : float
        Rotation angle in unit determined by "rotation_unit".
    rotation_unit : str
        'radians' or 'degrees' for the rotation angle.
    Returns:
    np.ndarray
        Values of the half-ellipse function at (X, Y).
    """
    if rotation:
        if rotation_unit == 'degrees':
            rotation = np.deg2rad(rotation)
        X_rot = np.cos(rotation) * X + np.sin(rotation) * Y
        Y_rot = -np.sin(rotation) * X + np.cos(rotation) * Y
        X = X_rot
        Y = Y_rot
    return (X**2) / ((X>=0)*(a**2)+(X<0)*b**2)  +  (Y**2) / (b**2)



plt.figure(figsize=(8, 6.5))
a=1.5
b=1
rotation=-np.pi/8
half_ellipse_values = half_ellipse(X,Y, a=a, b=b, rotation=rotation)
plot = plt.pcolormesh(X,Y,half_ellipse_values, alpha=1)
plt.scatter(X,Y, s=2, color='gray', alpha=0.2)
plt.scatter(X[half_ellipse_values<=1],Y[half_ellipse_values<=1], s=2, color='brown', alpha=0.8, label='inside half-ellipse')
plt.contour(X, Y, half_ellipse_values, levels=[1], alpha=0.7, linewidths=1.5, colors='black')
plt.axis('equal')
plt.legend(loc='upper right')
plt.title(f"a={a}, b={b}, rotation={rotation:.4f} (rad)")
plt.text(0,-1.8,"$\\frac{x_r^2}{0.5(sign(x_r)+1)*a^2 + 0.5(-sign(x_r)+1)*b^2} + \\frac{y_r^2}{b^2} = 1$", color='white')
plt.grid(alpha=0.3)
plt.colorbar(plot)
plt.show()
```

```python
def wind_to_angle_and_magnitude(wind_vector):
    wind_magnitude = np.sqrt(wind_vector[0]**2 + wind_vector[1]**2)
    wind_angle = np.arctan2(wind_vector[1], wind_vector[0])
    return wind_angle, wind_magnitude


plt.figure(figsize=(8,8))

wind_vector=(0.5,0.8)
wind_angle, wind_magnitude = wind_to_angle_and_magnitude(wind_vector)

half_ellipse_values = half_ellipse(X,Y,a=b+wind_magnitude,b=b, rotation=wind_angle)
plt.grid(alpha=0.5)
plt.contour(X, Y, half_ellipse_values, levels=[1], alpha=0.7, linewidths=1.5, colors='black')
plt.arrow(0,0, wind_vector[0], wind_vector[1], linewidth=3)
plt.title(f"Wind vector {wind_vector}, magnitude={wind_magnitude:.3f}, angle={wind_angle:.3f}")
plt.show()
```

```python
def general_ellipse(X,Y, x0=0,y0=0, a=2, b=1, rotation=0, rotation_unit='radians'):
    """

    """
    if rotation:
        if rotation_unit == 'degrees':
            rotation = np.deg2rad(rotation)
    A = a**2 * np.sin(rotation)**2 + b**2 * np.cos(rotation)**2
    B = 2*(b**2 - a**2) * np.sin(rotation) * np.cos(rotation)
    C = a**2 * np.cos(rotation)**2 + b**2 * np.sin(rotation)**2
    D = -2*A*x0 - B*y0
    E = -B*x0 - 2*C*y0
    F = A*x0**2 + B*x0*y0 + C*y0**2 - a**2 * b**2

    return A*X**2 + B*X*Y + C*Y**2 + D*X + E*Y + F

```
